```python
# Install library
!pip install google-play-scraper transformers torch pandas matplotlib langdetect wordcloud -q

print("✅ Semua library berhasil diinstall!")

# Import library
from google_play_scraper import reviews, Sort
import pandas as pd
import matplotlib.pyplot as plt
from transformers import pipeline
from langdetect import detect
from wordcloud import WordCloud

# =========================
# SCRAPING DATA GOOGLE PLAY
# =========================

APP_ID = 'com.Info_BMKG'

result, _ = reviews(
    APP_ID,
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=100
)

# Ambil data penting
data = []

for review in result:
    data.append({
        'userName': review['userName'],
        'score': review['score'],
        'at': review['at'],
        'content': review['content']
    })

# Buat DataFrame
df = pd.DataFrame(data)

# Simpan ulasan asli
df.to_csv('ulasan_google_play.csv', index=False)

print("✅ Data ulasan berhasil diambil!")
print(df.head())

# =========================
# ANALISIS SENTIMEN
# =========================

print("⏳ Memuat model sentiment...")

classifier = pipeline(
    "sentiment-analysis",
    model="w11wo/indonesian-roberta-base-sentiment-classifier"
)

# Fungsi sentiment
def get_sentiment(text):
    try:
        result = classifier(str(text[:512]))[0]
        return result['label']
    except:
        return 'neutral'

# Analisis sentiment
df['sentiment'] = df['content'].apply(get_sentiment)

print("✅ Analisis sentimen selesai!")

# =========================
# DETEKSI GAYA BAHASA
# =========================

def detect_language(text):
    try:
        lang = detect(str(text))

        if lang == 'id':
            return 'Indonesia'
        elif lang == 'en':
            return 'Inggris'
        elif lang == 'su':
            return 'Sunda'
        elif lang == 'jv':
            return 'Jawa'
        else:
            return 'Lainnya'

    except:
        return 'Tidak Diketahui'

df['gaya_bahasa'] = df['content'].apply(detect_language)

print("✅ Deteksi gaya bahasa selesai!")

# =========================
# DETEKSI TOPIK KOMENTAR
# =========================

def detect_topic(text):

    text = str(text).lower()

    if 'gempa' in text or 'cuaca' in text:
        return 'Informasi BMKG'

    elif 'error' in text or 'bug' in text:
        return 'Error/Bug'

    elif 'bagus' in text or 'mantap' in text:
        return 'Pujian'

    elif 'lemot' in text or 'loading' in text:
        return 'Performa'

    else:
        return 'Lainnya'

df['topik_komentar'] = df['content'].apply(detect_topic)

# =========================
# SIMPAN HASIL
# =========================

df.to_csv('hasil_sentimen.csv', index=False)
df.to_csv('hasil_analisis_lengkap.csv', index=False)

print("✅ File CSV berhasil disimpan!")

# =========================
# GRAFIK SENTIMEN
# =========================

sentiment_counts = df['sentiment'].value_counts()

plt.figure(figsize=(7,5))
sentiment_counts.plot(kind='bar')

plt.title('Grafik Sentimen Komentar')
plt.xlabel('Sentimen')
plt.ylabel('Jumlah')

plt.savefig('grafik_sentimen.png')
plt.show()

# =========================
# GRAFIK GAYA BAHASA
# =========================

bahasa_counts = df['gaya_bahasa'].value_counts()

plt.figure(figsize=(7,5))
bahasa_counts.plot(kind='bar')

plt.title('Grafik Gaya Bahasa')
plt.xlabel('Bahasa')
plt.ylabel('Jumlah')

plt.savefig('grafik_gaya_bahasa.png')
plt.show()

# =========================
# WORDCLOUD
# =========================

text = " ".join(df['content'].astype(str))

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white'
).generate(text)

plt.figure(figsize=(10,5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')

plt.title('WordCloud Komentar')

plt.savefig('wordcloud_komentar.png')
plt.show()

print("✅ Semua proses selesai!")
```
